## **1. Print every element inside this nested list:**

## **2. Find the sum of all numbers inside a nested list.**
Example: `[[10,20], [30], [5, 7, 8]]`

````python
# Write your answer here
l2 = [[10,20], [30], [5, 7, 8]]

res = 0
for i in l2:
    for j in i:
       res += j

print(res)
````

## **8. Flatten a nested list into a single list.**

## **9. Replace all negative numbers with 0 inside a nested list.**

````python
# Write your answer here
lst = [[10, -5, 20], [-1, 0, 50], [5, -10, -2, 15]]

res = [[num if num > 0 else 0 for num in res] for res in lst]

print(res)
````

## **14. Count how many students scored more than 80 in math:**
```
{
  'A': {'math': 85, 'science': 90},
  'B': {'math': 70, 'science': 88}
}
```

````python
# Write your answer here
dic = {
  'A': {'math': 85, 'science': 90},
  'B': {'math': 70, 'science': 88},
  'C': {'math': 90, 'science': 60}
}

count = 0
for v in dic.values():
    for k, j in v.items():
        if k == 'math' and j > 80:
            count += 1
print(count)
    
````
