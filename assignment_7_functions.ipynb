### Write a function process_students(data) that prints total marks and grade for each student.
- Example: [ {"name":"A", "marks":[80,90,70]}, {"name":"B", "marks":[60,75,65]} ] 
- Expected Output: Name: A | Total: 240 | Grade: A Name: B | Total: 200 | Grade: B

```python
def process_students(data):
    for student in data:
        name = student['name']
        marks = student['marks']

        total = sum(marks)

        if total >= 240:
            grade = 'A'
        elif total >= 200:
            grade = 'B'
        else:
            grade = 'C'
            
        print(f"Name: {name} | Total: {total} | Grade: {grade}")


data =  [{"name":"A", "marks":[80,90,70]}, {"name":"B", "marks":[60,75,65]} ]
process_students(data)
```

### Create a function apply_discounts(products) that updates price after discount. 
- Example: [ {"name":"Laptop", "price":50000, "discount":10}, {"name":"Phone", "price":20000, "discount":5} ] 
- Expected Output: Laptop final: 45000 Phone final: 19000

```python

def apply_discounts(products):
    for prod in products:
        name = prod["name"]
        price = prod["price"]
        discount = prod["discount"]

        updated_price = price -((price*discount)/100)
        #print(updated_price)
        print(f"{name} Final: {updated_price}")

products = [ {"name":"Laptop", "price":50000, "discount":10}, {"name":"Phone", "price":20000, "discount":5} ]

apply_discounts(products)
```

### Write a function average_salary(emp_list) that returns the average salary.

```python

def average_salary(emp_list):
    
    total = sum(emp["salary"] for emp in emp_list)
    avg = total/len(emp_list)
    print(avg)

emp_list = [
    {'name': 'Alice', 'salary': 60000},
    {'name': 'Bob', 'salary': 75000},
    {'name': 'Charlie', 'salary': 50000},
    {'name': 'David', 'salary': 90000}]

average_salary(emp_list)
```

### Write a function salary_hike(emp_list, percent) that increases each salary by given percent.

```python
def salary_hike(emp_list, percent):
    for emp in emp_list:
        name = emp["name"]
        salary = emp["salary"]

        updated_salary = salary + ((salary*percent)/100)
        print(updated_salary)

emp_list = [
    {'name': 'Alice', 'salary': 60000},
    {'name': 'Bob', 'salary': 75000},
    {'name': 'Charlie', 'salary': 50000},
    {'name': 'David', 'salary': 90000}]
percent = 10
salary_hike(emp_list, percent)
```
