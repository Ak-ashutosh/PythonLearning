## Q1: Create a class called Student

Requirements:
- Class variable: `school_name = "ABC International School"`
- Constructor: takes `name` and `marks`
- Instance method: `display_details()`
- Class method: `change_school()`
- Static method: `is_pass(marks)` returns "Pass" if marks ≥ 35 else "Fail"


```python

class Student:
    school_name = "ABC International School"

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display_details(self):
        print(f"Instance Method -> {self.name} has {self.marks} marks so student is {Student.is_pass(self.marks)}")

    @classmethod
    def change_school(cls, new_name):
        cls.school_name = new_name
        print(f"Class Method has -> {cls.school_name}")

    @staticmethod
    def is_pass(marks):
        if marks >= 35:
            return "Pass"
        else:
            return "Fail"
        

obj = Student("ASHU", 45)
obj.display_details()

obj.change_school("XYZ School")
```

## Q2: Create a class Calculator

Requirements:
- Class variable: `count_operations = 0`
- Instance method: `add(a, b)`
- Class method: `increment_count()`
- Static method: `info()`


```python
class Calculator:
    count_operations = 0

    def add(self, a, b):
        result = a+b
        self.increment_count()
        return result

    @classmethod
    def increment_count(cls):
        cls.count_operations += 1

    @staticmethod
    def info():
        return "This is a Calculator class which performed count operations."

# Create an instance
my_calc = Calculator()

# Perform operations
print(my_calc.add(5, 10))  # Output: 15
#print(my_calc.add(20, 30)) # Output: 50

# Check total operations across the whole class
print(Calculator.count_operations) # Output: 2

# Call the static info method
print(Calculator.info())
    
```

## Q3: Write a class BankAccount

Requirements:
- Class variable: `bank_name = "SBI"`
- Constructor: `name`, `balance`
- Instance method: `deposit(amount)`
- Class method: `change_bank(new_name)`
- Static method: `validate_amount(amount)`


```python
class BankAccount:
    bank_name = "SBI"

    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        if self.validate_amount(amount):
            self.balance += amount
            print(f"{amount} has credited to {self.balance}") 
        else:
            print("Validation Failed")

    @classmethod 
    def change_bank(cls, new_name):
        cls.bank_name = new_name
        print(f"New Bank name -> {cls.bank_name}")

    @staticmethod
    def validate_amount(amount):
        return amount > 0

account1 = BankAccount("ABC", 4500)

BankAccount.validate_amount(500)  # True

account1.deposit(500)

BankAccount.change_bank("State Bank of India")
account1.bank_name  
```
